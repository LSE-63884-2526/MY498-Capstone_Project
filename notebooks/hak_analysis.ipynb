## Robustness 2: Validating Masking + Whitening 
### ***through the Herfindahl-Hirschman Index***

I measure how concentrated the *hakkını* tokens are across clusters. A raw count of clusters is sensitive to k, so I use the **Herfindahl-Hirschman Index (HHI)**, which is independent of k.

**HHI**
$$HHI = \sum_{i=1}^{k} p_i^2, \quad p_i = \frac{n_i}{\sum_j n_j}$$

**Effective number of clusters**
$$N_{\text{eff}} = \frac{1}{HHI}$$

Here $p_i$ is the share of *hakkını* tokens in cluster $i$. HHI ranges from $1/k$ (fully dispersed) to $1$ (all in one cluster). Lower HHI and higher $N_{\text{eff}}$ mean the syntactic uses of *hak* have separated.

| Metric | Normal (a) | Masked + Whitening (b) | Change |
|---|---:|---:|---:|
| Total *hakkını* tokens | 1945 | 1681 | - |
| Clusters with >0 tokens | 57 | 97 | up |
| **HHI** | 0.072 | 0.032 | -56% |
| **Effective clusters** | 13.9 | 31.6 | +127% |
| Clusters covering 90% | 19 | 42 | up |

$N_{\text{eff}}$ more than doubles while k barely changes (108 vs 114). The dispersal comes from the signal processing, not from a larger k.

> Hirschman, A. O. (1945). *National Power and the Structure of Foreign Trade*. University of California Press. Inverse form: Simpson, E. H. (1949). Measurement of diversity. *Nature*, 163(4148), 688.
