```python
total_clustered = (millet.m.cluster != -1).sum()
assigned = set()
for name in millet.cluster_groups:
    assigned |= set(millet.cluster_groups[name])
in_meta = millet.m.cluster.isin(assigned).sum()

print("HDBSCAN clustered:", total_clustered)
print("assigned to meta-clusters:", in_meta)
print("excluded:", total_clustered - in_meta)
```
