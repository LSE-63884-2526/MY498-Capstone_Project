```python
total = halk.m[halk.m["cluster"] != -1]["cluster"].nunique()
grouped_ids = {cid for ids in halk.cluster_groups.values() for cid in ids}
grouped = len(grouped_ids)
total_occ = (halk.m["cluster"] != -1).sum()
grouped_occ = halk.m["cluster"].isin(grouped_ids).sum()
pct = 100 * grouped_occ / total_occ
print(f"{halk.term}: {grouped}/{total} clustered ({total-grouped} excluded), "
      f"occurrence included {pct:.0f}%")
```

```python
total_clustered = (halk.m.cluster != -1).sum()
assigned = set()
for name in halk.cluster_groups:
    assigned |= set(halk.cluster_groups[name])
in_meta = halk.m.cluster.isin(assigned).sum()

print("HDBSCAN clustered:", total_clustered)
print("assigned meta-clusters:", in_meta)
print("excluded:", total_clustered - in_meta)
```
