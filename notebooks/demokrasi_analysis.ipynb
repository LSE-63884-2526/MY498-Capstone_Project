```python
total = demokrasi.m[demokrasi.m["cluster"] != -1]["cluster"].nunique()
grouped_ids = {cid for ids in demokrasi.cluster_groups.values() for cid in ids}
grouped = len(grouped_ids)
total_occ = (demokrasi.m["cluster"] != -1).sum()
grouped_occ = demokrasi.m["cluster"].isin(grouped_ids).sum()
pct = 100 * grouped_occ / total_occ
print(f"{demokrasi.term}: {grouped}/{total} clusters grouped, ({total-grouped} excluded), "
      f"occurrence rate {pct:.0f}%")
```

```python
total_clustered = (demokrasi.m.cluster != -1).sum()
# assigned to meta clusters
assigned = set()
for name in demokrasi.cluster_groups:
    assigned |= set(demokrasi.cluster_groups[name])
in_meta = demokrasi.m.cluster.isin(assigned).sum()

print("HDBSCAN clustered:", total_clustered)
print("assigned to meta-clusters:", in_meta)
print("excluded:", total_clustered - in_meta)
```
